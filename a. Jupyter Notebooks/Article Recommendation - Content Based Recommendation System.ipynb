```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv("Article Dataset.csv",
                   encoding='latin1')
```

```python
print("Data read successfully.")
display(data.head(2))
display(data.tail(2))
```

```python
# Cosine similarity machine learning solution.
print("Extracting articles...")
articles = data["Article"].tolist()
print(f"Total articles extracted: {len(articles)}")
```

```python
print("Vectorizing the articles using TF-IDF...")
uni_tfidf = TfidfVectorizer(input='content', stop_words="english")
uni_matrix = uni_tfidf.fit_transform(articles)
print("Vectorization complete.")
```

```python
print("Calculating cosine similarity...")
uni_sim = cosine_similarity(uni_matrix)
print("Cosine similarity calculation complete.", uni_sim)
```

```python
def recommend_articles(x):
    return ", ".join(data["Title"].loc[x.argsort()[-5:-1]])
```

```python
print("Generating recommendations for each article...")
data["Recommended Articles"] = [recommend_articles(x) for x in uni_sim]
```

```python
print("Recommendations generated. Here are the first few rows with recommendations:")
display(data.head(5))
```

```python
print("First article's title and content:\n")
print(f"Title: {data['Title'][0]}\n")
print(f"Content: {data['Article'][0]}")
```

```python
print("Recommended articles for the first article:")
for article in data["Recommended Articles"][0].split(','):
    print("    -", article.lstrip())
```
